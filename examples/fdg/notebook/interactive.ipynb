## Failure, recovery and cleanup

Run `a.set_backend_available(False)` in a new cell: the real kernel rejects the next lease message and output A stops with a visible diagnostic. Run `a.set_backend_available(True)`, then click **Retry backend**. Output B remains independent. Clear outputs, rerun graph cells, or call `a.close(); b.close()` to dispose. Kernel restart stops old views when their comms close or the four-second lease expires; rerun the cells to recover.

Saved SVG is the initial settled frame. **Export current SVG** downloads the current real Moxi frame. A reopened notebook needs no live adapter to expand its saved SVG. No remote-kernel support is claimed from this local demonstration.
